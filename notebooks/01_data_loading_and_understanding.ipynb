```python
df['CREDIT_UTILIZATION'] = (
    df['BILL_AMT1'] / df['LIMIT_BAL']
) * 100

df['CREDIT_UTILIZATION'].describe().round(2)
```

```python
df['UTILIZATION_GROUP'] = pd.cut(
    df['CREDIT_UTILIZATION'],
    bins=[-float('inf'), 0, 25, 50, 75, 100, float('inf')],
    labels=['Negative', '0-25%', '25-50%', '50-75%', '75-100%', 'Above 100%']
)

utilization_default_rate = df.groupby(
    'UTILIZATION_GROUP',
    observed=False
)['default payment next month'].mean() * 100

utilization_default_rate.round(2)
```

```python
plt.figure(figsize=(9, 5))

utilization_default_rate.plot(
    kind='bar',
    edgecolor='black'
)

plt.title('Default Rate by Credit Utilisation Group')
plt.xlabel('Credit Utilisation Group')
plt.ylabel('Default Rate (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
correlation_columns = [
    'LIMIT_BAL',
    'AGE',
    'PAY_0',
    'PAY_2',
    'PAY_3',
    'BILL_AMT1',
    'BILL_AMT2',
    'PAY_AMT1',
    'PAY_AMT2',
    'default payment next month'
]

correlation_matrix = df[correlation_columns].corr()

correlation_matrix.round(2)
```

```python
plt.figure(figsize=(12, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap - Banking Credit Risk Analysis')
plt.tight_layout()
plt.show()
```

```python
default_correlation = correlation_matrix[
    'default payment next month'
].drop('default payment next month')

default_correlation = default_correlation.sort_values(
    key=abs,
    ascending=False
)

default_correlation.round(3)
```

```python
default_correlation.to_csv(
    '../reports/default_correlation_analysis.csv',
    header=['Correlation with Default']
)

print("Correlation analysis saved successfully!")
```

# Final Business Insights and Recommendations

## 1. Dataset Overview

- Total customer records: 30,000
- Total original dataset columns: 25
- Dataset source: UCI Machine Learning Repository
- Target variable: Default payment next month
- Non-default customers: 23,364 (77.88%)
- Default customers: 6,636 (22.12%)

The dataset contains historical credit card customer information, including demographics, credit limits, repayment status, billing amounts and payment amounts. It is used to analyse customer repayment behaviour and identify patterns associated with credit default.

## 2. Key Business Findings

1. **Default Distribution:** 22.12% of customers defaulted, while 77.88% did not default.

2. **Age Analysis:** Customers aged 61–79 recorded the highest default rate at 26.84%.

3. **Credit Limit Analysis:** Customers in the 0–50K credit limit group recorded a 31.79% default rate, compared with 11.17% in the 500K+ group.

4. **Gender Analysis:** Male customers recorded a 24.17% default rate, compared with 20.78% for female customers.

5. **Repayment Behaviour:** Customers with delayed repayment statuses generally showed higher default rates.

6. **Payment Analysis:** Non-default customers had higher average payment amounts across all six repayment periods.

7. **Credit Utilisation:** Customers with utilisation above 100% recorded the highest default rate at 30.07%.

8. **Correlation Analysis:** PAY_0 showed the strongest positive correlation with default at 0.325.

## 3. Business Recommendations

1. **Monitor Repayment Delays:** Banks should closely monitor customers with delayed repayment statuses to identify potential credit risks.

2. **Review High Credit Utilisation:** Customers using a high percentage of their available credit should receive additional attention during credit-risk assessments.

3. **Analyse Payment Behaviour:** Regularly reviewing customer payment amounts can help identify changes in repayment patterns.

4. **Improve Credit Risk Assessment:** Credit limits, repayment history and customer information should be considered together when evaluating credit risk.

5. **Early Risk Identification:** Historical customer behaviour can help banks identify patterns associated with default and support better risk-monitoring decisions.

## 4. Conclusion

This project analysed 30,000 historical credit card customer records to understand repayment behaviour and credit default patterns.

Python libraries such as Pandas, NumPy, Matplotlib and Seaborn were used for data cleaning, exploratory data analysis, statistical summaries and visualisation.

The analysis identified repayment status, credit limits and credit utilisation as useful areas for understanding customer default behaviour.

The findings can support credit-risk monitoring and help financial institutions make more informed decisions.
